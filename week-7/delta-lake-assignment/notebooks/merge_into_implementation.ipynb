### SCD Type 1

```python
%sql
MERGE INTO practice.default.customers AS cust
USING practice.default.customers_incremental AS cust_incr
ON cust.customer_id = cust_incr.customer_id
WHEN MATCHED THEN
    UPDATE SET
        name = cust_incr.name,
        email = cust_incr.email,
        city = cust_incr.city,
        signup_date = cust_incr.signup_date
WHEN NOT MATCHED THEN
    INSERT (customer_id, name, email, city, signup_date)
    VALUES (cust_incr.customer_id, cust_incr.name, cust_incr.email, cust_incr.city, cust_incr.signup_date)
```
